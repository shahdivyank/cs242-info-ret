```python
import numpy as np

times = np.load("../phase_1/indexing/times.npy")
stopwords = np.load("../phase_1/indexing/stopwords.npy")
bert_embeddings = np.load("./bert_embeddings_indexing.npy")
bert_embeddings_batches = np.load("./bert_embeddings_batches_indexing.npy")
bert_embeddings_batches_2000 = np.load("./bert_embeddings_batches_2000_indexing.npy")
bert_embeddings_batches_500 = np.load("./bert_embeddings_batches_500_indexing.npy")
```

```python
import matplotlib.pyplot as plt

TOTAL_DOCUMENTS = 137375

plt.figure(figsize=(8, 5))

plt.plot(range(TOTAL_DOCUMENTS), times, linestyle='-', label="ElasticSearch Naive Indexing")
plt.plot(range(TOTAL_DOCUMENTS), stopwords, linestyle='-', label="ElasticSearch Indexing Without Stop Words")
plt.plot(range(TOTAL_DOCUMENTS), bert_embeddings, linestyle='-', label="Bert Embeddings")
plt.plot(range(TOTAL_DOCUMENTS), bert_embeddings_batches, linestyle='-', label="Bert Embeddings (Batched B = 1000)")
plt.plot(range(TOTAL_DOCUMENTS), bert_embeddings_batches_2000, linestyle='-', label="Bert Embeddings (Batched B = 2000)")
plt.plot(range(TOTAL_DOCUMENTS), bert_embeddings_batches_500, linestyle='-', label="Bert Embeddings (Batched B = 500)")

plt.xlabel("Number of Documents")
plt.ylabel("Time (seconds)")
plt.title("Index Creation Performance")

plt.legend()
plt.show()
```
